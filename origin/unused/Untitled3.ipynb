```python
from pylab import *
```

```python
def H(w):
    wc = 4000*pi
    return 1.0 / (1.0 + 1j * w / wc)
```

```python
f = logspace(1,5) # frequencies from 10**1 to 10**5
```

```python
plot(f, 20*log10(abs(H(2*pi*f)))); xscale('log')
```

```python
from scipy import signal
```

```python
system = signal.lti([1], [1/(4000*pi),1])
```

```python
f = logspace(1, 5)
```

```python
w = 2 * pi * f
```

```python
w, mag, phase = signal.bode(system,w)
```

```python
semilogx( f, mag);
```

```python
semilogx(f, phase);
```
